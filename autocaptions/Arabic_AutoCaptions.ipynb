# كابشن عربي أوتوماتيك 🎬 Arabic Auto Captions

حط فيديو، وبيطلعلك نفس الفيديو مع **كابشن كلمات مفتاحية بالعربي** محروق عليه، مجانًا. بيفهم اللهجات، وبيحط الكلمة الأقوى بلون، وبيشيل «يعني» و«اممم»، وبيخلي الإنجليزي والأرقام بترتيبها الصحيح جوا الجملة العربية.

**الخطوات:**
1. من القائمة فوق: **Runtime ← Change runtime type ← T4 GPU** (أسرع بكثير، ومجاني).
2. اكبس ▶️ على كل خلية بالترتيب.
3. بآخر خلية بينزل الفيديو الجاهز، ومعه ملفات .srt و.ass إذا بدك تعدّل بـ CapCut أو Premiere.

Part of [Arabic Creator Kit](https://github.com/moathalherani/arabic-creator-kit) (MIT). Speech recognition: [faster-whisper](https://github.com/SYSTRAN/faster-whisper).

```python
#@title 1. تجهيز الأداة (مرة وحدة، حوالي دقيقة)
!pip -q install faster-whisper==1.2.1
!wget -q -O arabic_autocaptions.py https://raw.githubusercontent.com/moathalherani/arabic-creator-kit/main/autocaptions/arabic_autocaptions.py
import subprocess, torch
print('GPU ✓' if torch.cuda.is_available() else 'ما في GPU: الأداة بتشتغل بس أبطأ. غيّرها من Runtime ← Change runtime type')
print('جاهز ✓')
```

```python
#@title 2. ارفع الفيديو
from google.colab import files
uploaded = files.upload()
video = next(iter(uploaded))
print('الفيديو:', video)
```

```python
#@title 3. الإعدادات واعمل الكابشن
words_per_caption = 3 #@param {type:"slider", min:1, max:5, step:1}
font = "Changa" #@param ["Changa", "Noto Kufi Arabic", "Cairo", "Tajawal", "Kufam"]
text_color = "#FFFFFF" #@param {type:"string"}
keyword_color = "#7FB069" #@param {type:"string"}
model = "large-v3-turbo" #@param ["large-v3-turbo", "large-v3", "medium", "small"]
#@markdown كلمات بتحكيها وبدك الأداة تكتبها صح (أسماء، ماركات، كلمات إنجليزي):
hint_words = "" #@param {type:"string"}
keep_fillers = False #@param {type:"boolean"}

cmd = ['python', 'arabic_autocaptions.py', video, '--words', str(words_per_caption), '--font', font,
       '--color', text_color, '--highlight', keyword_color, '--model', model]
if hint_words.strip():
    cmd += ['--prompt', hint_words.strip()]
if keep_fillers:
    cmd += ['--keep-fillers']
subprocess.run(cmd, check=True)
```

```python
#@title 4. شوف النتيجة ونزّلها
import os, glob, zipfile
from IPython.display import HTML
from base64 import b64encode
stem = os.path.splitext(video)[0] + '.captioned'
out = stem + '.mp4'
if os.path.getsize(out) < 60_000_000:
    data = b64encode(open(out, 'rb').read()).decode()
    display(HTML(f'<video width=320 controls src="data:video/mp4;base64,{data}"></video>'))
with zipfile.ZipFile(stem + '.zip', 'w') as z:
    for f in glob.glob(stem + '.*'):
        if not f.endswith('.zip'):
            z.write(f)
files.download(out)
files.download(stem + '.zip')  # .srt و .ass و الكلمات بتوقيتها
```
